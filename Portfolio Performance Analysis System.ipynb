PERFORMANCE CLASSIFICATION

```python
#PERFORMANCE CLASSIFICATION

for asset in portfolio:
    quantity = asset["quantity"]
    purchase_price = asset["purchase_price"]
    current_price = asset["current_price"]
    transaction_cost = asset["transaction_cost"]
    dividend_income = asset["dividend_income"]
    initial_investment = quantity * purchase_price
    current_market_value = quantity * current_price


    profit_loss = (current_market_value - initial_investment - transaction_cost + dividend_income)


    return_percentage = (profit_loss / initial_investment) * 100


    if return_percentage >= 15:
        performance = "Excellent"

    elif return_percentage >= 5:
        performance = "Positive"

    elif return_percentage >= 0:
        performance = "Low Positive"

    else:
        performance = "Loss"


    print("--------------------------------")
    print("Asset:", asset["asset_name"])
    print(f"Initial Investment: {initial_investment:,}")
    print(f"Current Value: {current_market_value:,}")
    print(f"Profit/Loss: {profit_loss:,}")
    print(f"Return: {return_percentage:.2f}%")
    print("Performance:", performance)
```

PORTFOLIO TOTALS

```python
# PORTFOLIO TOTALS
total_investment = 0
total_current_value = 0
total_profit = 0

for asset in portfolio:
    quantity = asset["quantity"]
    initial_investment = (quantity * asset["purchase_price"])
    current_market_value = (quantity * asset["current_price"])
    profit_loss = (current_market_value - initial_investment - asset["transaction_cost"] + asset["dividend_income"])


    total_investment += initial_investment

    total_current_value += current_market_value

    total_profit += profit_loss

print(f"Total Investment: {total_investment:,}")
print(f"Total Current Value: {total_current_value:,}")
print(f"Total Profit: {total_profit:,}")
```

PORTFOLIO RETURN

```python
# PORTFOLIO RETURN

portfolio_return = (total_profit / total_investment) * 100

print(f"Portfolio Return: {portfolio_return:.2f}%")
```

Which asset performed best?

```python
# Which asset performed best?

best_asset = ""

best_return = 0

for asset in portfolio:
    initial_investment = asset["quantity"] * asset["purchase_price"]
    current_market_value = asset["quantity"] * asset["current_price"]

    profit_loss = (current_market_value - initial_investment - asset["transaction_cost"] + asset["dividend_income"])
    
    return_percentage = (profit_loss / initial_investment) * 100
    
    if return_percentage > best_return:
        best_return = return_percentage
        best_asset = asset["asset_name"]

print("Best Performing Asset:", best_asset)
print(f"Best Return: {best_return:.2f}%")
```

THE NESTED LOOP CHALLENGE

```python
#46. THE NESTED LOOP CHALLENGE
countries = ["Nigeria", "Ghana", "Kenya", "South Africa"]


for country in countries:
    country_total = 0

    for asset in portfolio:
        if asset["country"] == country:
            current_market_value = (asset["quantity"]* asset["current_price"])
            country_total += current_market_value

    print(country, ":", f"{country_total:,}")
```

Nested while loop

```python
countries = ["Nigeria", "Ghana", "Kenya", "South Africa"]

country_index = 0

while country_index < len(countries):
    country = countries[country_index]
    asset_index = 0
    country_total = 0

    while asset_index < len(portfolio):
        asset = portfolio[asset_index]

        if asset["country"] == country:
            current_market_value = asset["quantity"] * asset["current_price"]
            country_total += current_market_value

        asset_index += 1

    print(country, ":", f"{country_total:,}")
    country_index += 1
```
